```python
def build_joint_only_causal_mask(step_k_len: int, joint_len: int, *,
                                joint_strategy: str,
                                device, dtype=torch.bool):
    """
    Build a mask of shape [1, q_total, k_total] (True=allowed, False=masked),
    where q_total = step_k_len + joint_len and k_total = step_k_len + joint_len.

    Rules:
    - Queries from the step_k segment: no masking -> can attend to all keys (including joint).
    - Queries from the joint segment: apply causal masking only within the joint->joint block
      (i.e., joint-internal causality). All other blocks remain unmasked.
    """
    q_total = step_k_len + joint_len
    k_total = q_total
    mask = torch.ones((1, q_total, k_total), device=device, dtype=dtype)

    if joint_len == 0:
        return mask
        
    qj = torch.arange(joint_len, device=device)[:, None]  # [J, 1] query indices within joint
    kj = torch.arange(joint_len, device=device)[None, :]  # [1, J] key indices within joint
    joint_allow = (kj <= qj)  # [J, J]
    if joint_strategy == "rear":
        # Layout: [step_k, joint]
        # Joint query rows range: [step_k_len : step_k_len + joint_len)
        qs = ks = step_k_len
        qe = ke = step_k_len + joint_len
        mask[:, qs:qe, ks:ke] = joint_allow
        # Keep other blocks as True so step_k queries can attend to all joint keys.

    elif joint_strategy == "front":
        # Layout: [joint, step_k]
        # Joint query rows range: [0 : joint_len)
        mask[:, 0:joint_len, 0:joint_len] = joint_allow
        # Keep other blocks as True.
    else:
        raise ValueError(f"unknown joint_strategy: {joint_strategy}")

    return mask
```
